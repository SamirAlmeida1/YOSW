```python
!pip install wfdb
import numpy as np
from scipy.signal import butter, filtfilt, iirnotch
import wfdb
import os

# -----------------------------------------------------------------
# 1. CONFIGURACIÓN
# -----------------------------------------------------------------

FS = 360  # Frecuencia de muestreo de MIT-BIH (Hz)
LEAD = 'MLII'  # Derivación a usar

# Ventana de segmentación alrededor de cada R-peak (en muestras)
PRE_SAMPLES = 90    # ~0.25 s antes del pico R
POST_SAMPLES = 110  # ~0.30 s después del pico R

# Mapeo de símbolos MIT-BIH -> clases AAMI (N, S, V, F)
AAMI_MAP = {
    'N': 'N', 'L': 'N', 'R': 'N', 'e': 'N', 'j': 'N',
    'A': 'S', 'a': 'S', 'J': 'S', 'S': 'S',
    'V': 'V', 'E': 'V',
    'F': 'F',
}

RECORD_LIST = ['100', '101', '103', '105', '106', '108', '109', '111',
                '112', '113', '114', '115', '116', '117', '118', '119']

DATA_DIR = './mitbih_data'
OUTPUT_FILE = 'ecg_beats_dataset.npz'

# -----------------------------------------------------------------
# 2. DESCARGA DE REGISTROS
# -----------------------------------------------------------------

def descargar_registro(record_name, data_dir=DATA_DIR):
    """Descarga un registro completo de MIT-BIH si falta algún archivo esencial."""
    os.makedirs(data_dir, exist_ok=True)
    path = os.path.join(data_dir, record_name)

    # Verificar si faltan archivos críticos (.dat, .hea, .atr)
    extensiones = ['.dat', '.hea', '.atr']
    faltan_archivos = any(not os.path.exists(path + ext) for ext in extensiones)

    if faltan_archivos:
        print(f"Descargando registro {record_name} desde PhysioNet...")
        wfdb.dl_database('mitdb', dl_dir=data_dir, records=[record_name])
    return path

# -----------------------------------------------------------------
# 3. FILTRADO DE RUIDO
# -----------------------------------------------------------------

def filtrar_baseline_wander(signal, fs=FS, cutoff=0.5):
    nyquist = fs / 2
    b, a = butter(N=2, Wn=cutoff / nyquist, btype='highpass')
    return filtfilt(b, a, signal)

def filtrar_ruido_linea(signal, fs=FS, freq=60.0, quality=30.0):
    b, a = iirnotch(w0=freq / (fs / 2), Q=quality)
    return filtfilt(b, a, signal)

def preprocesar_senal(signal, fs=FS):
    signal = filtrar_baseline_wander(signal, fs)
    signal = filtrar_ruido_linea(signal, fs)
    return signal

# -----------------------------------------------------------------
# 4. NORMALIZACIÓN
# -----------------------------------------------------------------

def normalizar_zscore(signal):
    mean = np.mean(signal)
    std = np.std(signal)
    return (signal - mean) / std if std != 0 else signal - mean

# -----------------------------------------------------------------
# 5. SEGMENTACIÓN EN LATIDOS
# -----------------------------------------------------------------

def segmentar_latidos(signal, r_peaks, symbols, pre=PRE_SAMPLES, post=POST_SAMPLES):
    beats = []
    labels = []
    for peak, symbol in zip(r_peaks, symbols):
        if symbol not in AAMI_MAP: continue
        start, end = peak - pre, peak + post
        if start < 0 or end > len(signal): continue
        beats.append(signal[start:end])
        labels.append(AAMI_MAP[symbol])
    return np.array(beats), np.array(labels)

# -----------------------------------------------------------------
# 6. PIPELINE POR REGISTRO Y CONSTRUCCIÓN
# -----------------------------------------------------------------

def procesar_registro(record_name, data_dir=DATA_DIR, lead=LEAD):
    path = descargar_registro(record_name, data_dir)
    record = wfdb.rdrecord(path)
    annotation = wfdb.rdann(path, 'atr')

    if lead not in record.sig_name:
        print(f"  [aviso] {record_name}: derivación {lead} no encontrada")
        return None, None

    raw_signal = record.p_signal[:, record.sig_name.index(lead)]
    normalized = normalizar_zscore(preprocesar_senal(raw_signal))
    return segmentar_latidos(normalized, annotation.sample, annotation.symbol)

def construir_dataset(record_list=RECORD_LIST):
    all_beats, all_labels = [], []
    for record_name in record_list:
        print(f"Procesando registro {record_name}...")
        beats, labels = procesar_registro(record_name)
        if beats is not None:
            all_beats.append(beats)
            all_labels.append(labels)
            print(f"  -> {len(beats)} latidos extraídos")

    X, y = np.concatenate(all_beats, axis=0), np.concatenate(all_labels, axis=0)
    print(f"\nDataset final: {X.shape[0]} latidos")
    return X, y

if __name__ == '__main__':
    X, y = construir_dataset()
    np.savez(OUTPUT_FILE, X=X, y=y)
    print(f"Dataset guardado en {OUTPUT_FILE}")
```

```python
"""
Preprocesamiento de señales ECG - MIT-BIH Arrhythmia Database
================================================================
Días 3-4 del proyecto: Clasificación multiclase de arritmias

Derivación usada: MLII (una sola derivación)
Clases AAMI: N, S, V, F (se excluye Q)

Requisitos (instalar antes de correr):
    pip install wfdb numpy scipy matplotlib

Nota: este script descarga los registros directamente desde PhysioNet,
por lo que necesitas conexión a internet (correr en local o en Colab).
"""

import numpy as np
import wfdb
from scipy.signal import butter, filtfilt, iirnotch
import os

# -----------------------------------------------------------------
# 1. CONFIGURACIÓN
# -----------------------------------------------------------------

FS = 360  # Frecuencia de muestreo de MIT-BIH (Hz)
LEAD = 'MLII'  # Derivación a usar

# Ventana de segmentación alrededor de cada R-peak (en muestras)
PRE_SAMPLES = 90    # ~0.25 s antes del pico R
POST_SAMPLES = 110  # ~0.30 s después del pico R

# Mapeo de símbolos MIT-BIH -> clases AAMI (N, S, V, F)
# Se descartan símbolos que corresponderían a Q (paced, unknown, etc.)
AAMI_MAP = {
    # Clase N (Normal)
    'N': 'N', 'L': 'N', 'R': 'N', 'e': 'N', 'j': 'N',
    # Clase S (Supraventricular ectopic)
    'A': 'S', 'a': 'S', 'J': 'S', 'S': 'S',
    # Clase V (Ventricular ectopic)
    'V': 'V', 'E': 'V',
    # Clase F (Fusion)
    'F': 'F',
}
# Símbolos NO incluidos (se descartan): '/', 'f', 'Q', '!', '+', '~', etc.

# -----------------------------------------------------------------
# Partición inter-paciente (patient-independent split)
# Basada en de Chazal et al. (2004), estándar de facto en la literatura
# para evitar data leakage: ningún paciente aparece en ambos conjuntos.
# Se excluyen 102, 104, 107, 217 por contener latidos con marcapasos.
# -----------------------------------------------------------------

DS1_TRAIN = ['101', '106', '108', '109', '112', '114', '115', '116',
             '118', '119', '122', '124', '201', '203', '205', '207',
             '208', '209', '215', '220', '223', '230']

DS2_TEST = ['100', '103', '105', '111', '113', '117', '121', '123',
            '200', '202', '210', '212', '213', '214', '219', '221',
            '222', '228', '231', '232', '233', '234']

DATA_DIR = './mitbih_data'
OUTPUT_FILE_TRAIN = 'ecg_beats_train.npz'
OUTPUT_FILE_TEST = 'ecg_beats_test.npz'


# -----------------------------------------------------------------
# 2. DESCARGA DE REGISTROS
# -----------------------------------------------------------------

def descargar_registro(record_name, data_dir=DATA_DIR):
    """Descarga un registro de MIT-BIH desde PhysioNet si no existe localmente."""
    os.makedirs(data_dir, exist_ok=True)
    path = os.path.join(data_dir, record_name)
    if not os.path.exists(path + '.dat'):
        wfdb.dl_database('mitdb', dl_dir=data_dir, records=[record_name])
    return path


# -----------------------------------------------------------------
# 3. FILTRADO DE RUIDO
# -----------------------------------------------------------------

def filtrar_baseline_wander(signal, fs=FS, cutoff=0.5):
    """Filtro pasa-altos Butterworth para eliminar baseline wander."""
    nyquist = fs / 2
    b, a = butter(N=2, Wn=cutoff / nyquist, btype='highpass')
    return filtfilt(b, a, signal)


def filtrar_ruido_linea(signal, fs=FS, freq=60.0, quality=30.0):
    """Filtro notch para eliminar ruido de línea eléctrica (50 o 60 Hz)."""
    b, a = iirnotch(w0=freq / (fs / 2), Q=quality)
    return filtfilt(b, a, signal)


def preprocesar_senal(signal, fs=FS):
    """Aplica la cadena completa de filtrado de ruido."""
    signal = filtrar_baseline_wander(signal, fs)
    signal = filtrar_ruido_linea(signal, fs)
    return signal


# -----------------------------------------------------------------
# 4. NORMALIZACIÓN
# -----------------------------------------------------------------

def normalizar_zscore(signal):
    """Normalización z-score: media 0, desviación estándar 1."""
    mean = np.mean(signal)
    std = np.std(signal)
    if std == 0:
        return signal - mean
    return (signal - mean) / std


# -----------------------------------------------------------------
# 5. SEGMENTACIÓN EN LATIDOS
# -----------------------------------------------------------------

def segmentar_latidos(signal, r_peaks, symbols, pre=PRE_SAMPLES, post=POST_SAMPLES):
    """
    Extrae ventanas fijas alrededor de cada R-peak y mapea el símbolo
    a su clase AAMI correspondiente. Descarta latidos:
      - demasiado cerca del inicio/fin de la señal (ventana incompleta)
      - cuyo símbolo no está en el mapeo AAMI (p.ej. clase Q)
    """
    beats = []
    labels = []

    for peak, symbol in zip(r_peaks, symbols):
        if symbol not in AAMI_MAP:
            continue  # descarta símbolos fuera de N, S, V, F (incluye Q)

        start = peak - pre
        end = peak + post
        if start < 0 or end > len(signal):
            continue  # ventana incompleta al borde de la señal

        beat = signal[start:end]
        beats.append(beat)
        labels.append(AAMI_MAP[symbol])

    return np.array(beats), np.array(labels)


# -----------------------------------------------------------------
# 6. PIPELINE COMPLETO POR REGISTRO
# -----------------------------------------------------------------

def procesar_registro(record_name, data_dir=DATA_DIR, lead=LEAD):
    """Ejecuta el pipeline completo para un registro: carga, filtra,
    normaliza y segmenta en latidos etiquetados."""
    path = descargar_registro(record_name, data_dir)

    record = wfdb.rdrecord(path)
    annotation = wfdb.rdann(path, 'atr')

    # Seleccionar la derivación MLII
    if lead not in record.sig_name:
        print(f"  [aviso] {record_name}: derivación {lead} no encontrada, se omite")
        return None, None

    lead_idx = record.sig_name.index(lead)
    raw_signal = record.p_signal[:, lead_idx]

    # Preprocesamiento
    filtered = preprocesar_senal(raw_signal)
    normalized = normalizar_zscore(filtered)

    # Segmentación usando anotaciones del propio dataset (R-peaks reales)
    r_peaks = annotation.sample
    symbols = annotation.symbol

    beats, labels = segmentar_latidos(normalized, r_peaks, symbols)
    return beats, labels


# -----------------------------------------------------------------
# 7. CONSTRUCCIÓN DEL DATASET COMPLETO
# -----------------------------------------------------------------

def construir_dataset(record_list):
    all_beats = []
    all_labels = []

    for record_name in record_list:
        print(f"Procesando registro {record_name}...")
        beats, labels = procesar_registro(record_name)
        if beats is None:
            continue
        all_beats.append(beats)
        all_labels.append(labels)
        print(f"  -> {len(beats)} latidos extraídos")

    X = np.concatenate(all_beats, axis=0)
    y = np.concatenate(all_labels, axis=0)

    print(f"\nDataset final: {X.shape[0]} latidos, longitud de ventana {X.shape[1]}")
    print("Distribución de clases:")
    for clase in np.unique(y):
        n = np.sum(y == clase)
        print(f"  {clase}: {n} ({100*n/len(y):.2f}%)")

    return X, y


if __name__ == '__main__':
    print("========== CONSTRUYENDO CONJUNTO DE ENTRENAMIENTO (DS1) ==========")
    X_train, y_train = construir_dataset(DS1_TRAIN)
    np.savez(OUTPUT_FILE_TRAIN, X=X_train, y=y_train)
    print(f"\nDataset de entrenamiento guardado en {OUTPUT_FILE_TRAIN}")

    print("\n========== CONSTRUYENDO CONJUNTO DE PRUEBA (DS2) ==========")
    X_test, y_test = construir_dataset(DS2_TEST)
    np.savez(OUTPUT_FILE_TEST, X=X_test, y=y_test)
    print(f"\nDataset de prueba guardado en {OUTPUT_FILE_TEST}")

    print("\n========== RESUMEN FINAL ==========")
    print(f"Train: {X_train.shape[0]} latidos | Test: {X_test.shape[0]} latidos")
    print(f"Total de pacientes: {len(DS1_TRAIN)} en train, {len(DS2_TEST)} en test (sin solapamiento)")
```

```python
"""
Primer modelo: CNN 1D para clasificación de arritmias
========================================================
Días 5-6 del proyecto: Clasificación multiclase de arritmias

Carga los datasets generados en el preprocesamiento (ecg_beats_train.npz,
ecg_beats_test.npz), entrena una CNN 1D y evalúa con las métricas
estándar: accuracy, precision, recall, F1-score y matriz de confusión.

Requisitos:
    pip install tensorflow scikit-learn numpy matplotlib
"""

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (accuracy_score, precision_recall_fscore_support,
                                classification_report, confusion_matrix,
                                ConfusionMatrixDisplay)

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks

# -----------------------------------------------------------------
# 1. CARGA DE DATOS
# -----------------------------------------------------------------

TRAIN_FILE = 'ecg_beats_train.npz'
TEST_FILE = 'ecg_beats_test.npz'

def cargar_datos(train_file=TRAIN_FILE, test_file=TEST_FILE):
    train_data = np.load(train_file)
    test_data = np.load(test_file)
    X_train, y_train = train_data['X'], train_data['y']
    X_test, y_test = test_data['X'], test_data['y']
    return X_train, y_train, X_test, y_test


# -----------------------------------------------------------------
# 2. PREPARACIÓN DE ETIQUETAS Y FORMA DE ENTRADA
# -----------------------------------------------------------------

def preparar_datos(X_train, y_train, X_test, y_test):
    # Codificar etiquetas de texto ('N','S','V','F') a enteros (0,1,2,3)
    encoder = LabelEncoder()
    y_train_enc = encoder.fit_transform(y_train)
    y_test_enc = encoder.transform(y_test)

    # CNN 1D espera forma (n_muestras, n_pasos_tiempo, n_canales)
    X_train_r = X_train.reshape(-1, X_train.shape[1], 1)
    X_test_r = X_test.reshape(-1, X_test.shape[1], 1)

    return X_train_r, y_train_enc, X_test_r, y_test_enc, encoder


# -----------------------------------------------------------------
# 3. ARQUITECTURA: CNN 1D
# -----------------------------------------------------------------

def construir_cnn(input_shape, n_clases):
    model = models.Sequential([
        layers.Input(shape=input_shape),

        layers.Conv1D(32, kernel_size=5, activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling1D(pool_size=2),

        layers.Conv1D(64, kernel_size=5, activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling1D(pool_size=2),

        layers.Conv1D(128, kernel_size=3, activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling1D(pool_size=2),

        layers.GlobalAveragePooling1D(),
        layers.Dense(64, activation='relu'),
        layers.Dropout(0.4),
        layers.Dense(n_clases, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model


# -----------------------------------------------------------------
# 4. PESOS DE CLASE (para mitigar el desbalance N vs S/V/F)
# -----------------------------------------------------------------

def calcular_pesos_clase(y_train_enc):
    clases = np.unique(y_train_enc)
    pesos = compute_class_weight(class_weight='balanced', classes=clases, y=y_train_enc)
    return dict(zip(clases, pesos))


# -----------------------------------------------------------------
# 5. ENTRENAMIENTO
# -----------------------------------------------------------------

def entrenar_modelo(model, X_train, y_train, class_weights, epochs=30, batch_size=128):
    early_stop = callbacks.EarlyStopping(
        monitor='val_loss', patience=5, restore_best_weights=True
    )

    history = model.fit(
        X_train, y_train,
        validation_split=0.15,
        epochs=epochs,
        batch_size=batch_size,
        class_weight=class_weights,
        callbacks=[early_stop],
        verbose=2
    )
    return history


# -----------------------------------------------------------------
# 6. EVALUACIÓN
# -----------------------------------------------------------------

def evaluar_modelo(model, X_test, y_test, encoder):
    y_pred_probs = model.predict(X_test)
    y_pred = np.argmax(y_pred_probs, axis=1)

    acc = accuracy_score(y_test, y_pred)
    precision, recall, f1, _ = precision_recall_fscore_support(
        y_test, y_pred, average='macro', zero_division=0
    )

    print(f"\n{'='*50}")
    print("RESULTADOS EN CONJUNTO DE PRUEBA (DS2)")
    print(f"{'='*50}")
    print(f"Accuracy:           {acc:.4f}")
    print(f"Precision (macro):  {precision:.4f}")
    print(f"Recall (macro):     {recall:.4f}")
    print(f"F1-score (macro):   {f1:.4f}")

    print("\nReporte por clase:")
    print(classification_report(y_test, y_pred, target_names=encoder.classes_,
                                   zero_division=0))

    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=encoder.classes_)
    fig, ax = plt.subplots(figsize=(6, 6))
    disp.plot(ax=ax, cmap='Blues', values_format='d')
    plt.title('Matriz de Confusión - CNN 1D')
    plt.tight_layout()
    plt.savefig('confusion_matrix_cnn.png', dpi=100)
    print("\nMatriz de confusión guardada en confusion_matrix_cnn.png")

    return {'accuracy': acc, 'precision': precision, 'recall': recall, 'f1': f1}


def graficar_historial(history):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(history.history['loss'], label='train')
    axes[0].plot(history.history['val_loss'], label='val')
    axes[0].set_title('Pérdida (Loss)')
    axes[0].legend()

    axes[1].plot(history.history['accuracy'], label='train')
    axes[1].plot(history.history['val_accuracy'], label='val')
    axes[1].set_title('Accuracy')
    axes[1].legend()

    plt.tight_layout()
    plt.savefig('training_history_cnn.png', dpi=100)
    print("Historial de entrenamiento guardado en training_history_cnn.png")


# -----------------------------------------------------------------
# 7. PIPELINE COMPLETO
# -----------------------------------------------------------------

if __name__ == '__main__':
    print("Cargando datos...")
    X_train, y_train, X_test, y_test = cargar_datos()

    print(f"Train: {X_train.shape[0]} latidos | Test: {X_test.shape[0]} latidos")

    X_train, y_train_enc, X_test, y_test_enc, encoder = preparar_datos(
        X_train, y_train, X_test, y_test
    )
    print(f"Clases (orden): {list(encoder.classes_)}")

    class_weights = calcular_pesos_clase(y_train_enc)
    print(f"Pesos de clase (para mitigar desbalance): {class_weights}")

    model = construir_cnn(input_shape=(X_train.shape[1], 1), n_clases=len(encoder.classes_))
    model.summary()

    history = entrenar_modelo(model, X_train, y_train_enc, class_weights)
    graficar_historial(history)

    resultados = evaluar_modelo(model, X_test, y_test_enc, encoder)

    model.save('modelo_cnn_arritmias.keras')
    print("\nModelo guardado en modelo_cnn_arritmias.keras")
```

```python
"""
Primer modelo: CNN 1D para clasificación de arritmias
========================================================
Días 5-6 del proyecto: Clasificación multiclase de arritmias

Carga los datasets generados en el preprocesamiento (ecg_beats_train.npz,
ecg_beats_test.npz), entrena una CNN 1D y evalúa con las métricas
estándar: accuracy, precision, recall, F1-score y matriz de confusión.

Requisitos:
    pip install tensorflow scikit-learn numpy matplotlib
"""

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (accuracy_score, precision_recall_fscore_support,
                                classification_report, confusion_matrix,
                                ConfusionMatrixDisplay)

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks

# -----------------------------------------------------------------
# 1. CARGA DE DATOS
# -----------------------------------------------------------------

TRAIN_FILE = 'ecg_beats_train.npz'
TEST_FILE = 'ecg_beats_test.npz'

def cargar_datos(train_file=TRAIN_FILE, test_file=TEST_FILE):
    train_data = np.load(train_file)
    test_data = np.load(test_file)
    X_train, y_train = train_data['X'], train_data['y']
    X_test, y_test = test_data['X'], test_data['y']
    return X_train, y_train, X_test, y_test


# -----------------------------------------------------------------
# 2. PREPARACIÓN DE ETIQUETAS Y FORMA DE ENTRADA
# -----------------------------------------------------------------

def preparar_datos(X_train, y_train, X_test, y_test):
    # Codificar etiquetas de texto ('N','S','V','F') a enteros (0,1,2,3)
    encoder = LabelEncoder()
    y_train_enc = encoder.fit_transform(y_train)
    y_test_enc = encoder.transform(y_test)

    # CNN 1D espera forma (n_muestras, n_pasos_tiempo, n_canales)
    X_train_r = X_train.reshape(-1, X_train.shape[1], 1)
    X_test_r = X_test.reshape(-1, X_test.shape[1], 1)

    return X_train_r, y_train_enc, X_test_r, y_test_enc, encoder


# -----------------------------------------------------------------
# 3. ARQUITECTURA: CNN 1D
# -----------------------------------------------------------------

def construir_cnn(input_shape, n_clases):
    model = models.Sequential([
        layers.Input(shape=input_shape),

        layers.Conv1D(32, kernel_size=5, activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling1D(pool_size=2),

        layers.Conv1D(64, kernel_size=5, activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling1D(pool_size=2),

        layers.Conv1D(128, kernel_size=3, activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling1D(pool_size=2),

        layers.GlobalAveragePooling1D(),
        layers.Dense(64, activation='relu'),
        layers.Dropout(0.4),
        layers.Dense(n_clases, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model


# -----------------------------------------------------------------
# 4. PESOS DE CLASE (para mitigar el desbalance N vs S/V/F)
# -----------------------------------------------------------------

def calcular_pesos_clase(y_train_enc):
    clases = np.unique(y_train_enc)
    pesos = compute_class_weight(class_weight='balanced', classes=clases, y=y_train_enc)
    return dict(zip(clases, pesos))


# -----------------------------------------------------------------
# 5. ENTRENAMIENTO
# -----------------------------------------------------------------

def entrenar_modelo(model, X_train, y_train, class_weights, epochs=30, batch_size=128):
    # IMPORTANTE: validation_split de Keras toma el último X% del arreglo TAL
    # CUAL, sin mezclar. Como X_train está ordenado por registro (paciente),
    # eso generaba una validación no representativa (a veces 1-2 pacientes).
    # Aquí usamos un split aleatorio Y ESTRATIFICADO (mantiene la proporción
    # de cada clase en train y validación).
    X_tr, X_val, y_tr, y_val = train_test_split(
        X_train, y_train, test_size=0.15, stratify=y_train, random_state=42
    )

    early_stop = callbacks.EarlyStopping(
        monitor='val_loss', patience=5, restore_best_weights=True
    )

    history = model.fit(
        X_tr, y_tr,
        validation_data=(X_val, y_val),
        epochs=epochs,
        batch_size=batch_size,
        class_weight=class_weights,
        callbacks=[early_stop],
        verbose=2
    )
    return history


# -----------------------------------------------------------------
# 6. EVALUACIÓN
# -----------------------------------------------------------------

def evaluar_modelo(model, X_test, y_test, encoder):
    y_pred_probs = model.predict(X_test)
    y_pred = np.argmax(y_pred_probs, axis=1)

    acc = accuracy_score(y_test, y_pred)
    precision, recall, f1, _ = precision_recall_fscore_support(
        y_test, y_pred, average='macro', zero_division=0
    )

    print(f"\n{'='*50}")
    print("RESULTADOS EN CONJUNTO DE PRUEBA (DS2)")
    print(f"{'='*50}")
    print(f"Accuracy:           {acc:.4f}")
    print(f"Precision (macro):  {precision:.4f}")
    print(f"Recall (macro):     {recall:.4f}")
    print(f"F1-score (macro):   {f1:.4f}")

    print("\nReporte por clase:")
    print(classification_report(y_test, y_pred, target_names=encoder.classes_,
                                   zero_division=0))

    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=encoder.classes_)
    fig, ax = plt.subplots(figsize=(6, 6))
    disp.plot(ax=ax, cmap='Blues', values_format='d')
    plt.title('Matriz de Confusión - CNN 1D')
    plt.tight_layout()
    plt.savefig('confusion_matrix_cnn.png', dpi=100)
    print("\nMatriz de confusión guardada en confusion_matrix_cnn.png")

    return {'accuracy': acc, 'precision': precision, 'recall': recall, 'f1': f1}


def graficar_historial(history):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(history.history['loss'], label='train')
    axes[0].plot(history.history['val_loss'], label='val')
    axes[0].set_title('Pérdida (Loss)')
    axes[0].legend()

    axes[1].plot(history.history['accuracy'], label='train')
    axes[1].plot(history.history['val_accuracy'], label='val')
    axes[1].set_title('Accuracy')
    axes[1].legend()

    plt.tight_layout()
    plt.savefig('training_history_cnn.png', dpi=100)
    print("Historial de entrenamiento guardado en training_history_cnn.png")


# -----------------------------------------------------------------
# 7. PIPELINE COMPLETO
# -----------------------------------------------------------------

if __name__ == '__main__':
    print("Cargando datos...")
    X_train, y_train, X_test, y_test = cargar_datos()

    print(f"Train: {X_train.shape[0]} latidos | Test: {X_test.shape[0]} latidos")

    X_train, y_train_enc, X_test, y_test_enc, encoder = preparar_datos(
        X_train, y_train, X_test, y_test
    )
    print(f"Clases (orden): {list(encoder.classes_)}")

    class_weights = calcular_pesos_clase(y_train_enc)
    print(f"Pesos de clase (para mitigar desbalance): {class_weights}")

    model = construir_cnn(input_shape=(X_train.shape[1], 1), n_clases=len(encoder.classes_))
    model.summary()

    history = entrenar_modelo(model, X_train, y_train_enc, class_weights)
    graficar_historial(history)

    resultados = evaluar_modelo(model, X_test, y_test_enc, encoder)

    model.save('modelo_cnn_arritmias.keras')
    print("\nModelo guardado en modelo_cnn_arritmias.keras")
```

```python
"""
Utilidades compartidas para el proyecto de clasificación de arritmias.
========================================================================
Usado por los scripts de cada arquitectura (CNN, MLP, LSTM, CNN-LSTM)
para garantizar que TODAS se evalúen bajo el mismo protocolo exacto:
mismo split train/val, mismas métricas, mismo formato de reporte.
"""

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (accuracy_score, precision_recall_fscore_support,
                                classification_report, confusion_matrix,
                                ConfusionMatrixDisplay)
from tensorflow.keras import callbacks

TRAIN_FILE = 'ecg_beats_train.npz'
TEST_FILE = 'ecg_beats_test.npz'


def cargar_datos(train_file=TRAIN_FILE, test_file=TEST_FILE):
    train_data = np.load(train_file)
    test_data = np.load(test_file)
    return train_data['X'], train_data['y'], test_data['X'], test_data['y']


def preparar_datos(X_train, y_train, X_test, y_test, forma='secuencia'):
    """
    forma='secuencia' -> (n, 200, 1)  para CNN/LSTM/CNN-LSTM
    forma='plana'      -> (n, 200)     para MLP
    """
    encoder = LabelEncoder()
    y_train_enc = encoder.fit_transform(y_train)
    y_test_enc = encoder.transform(y_test)

    if forma == 'secuencia':
        X_train_r = X_train.reshape(-1, X_train.shape[1], 1)
        X_test_r = X_test.reshape(-1, X_test.shape[1], 1)
    else:  # plana
        X_train_r = X_train.reshape(X_train.shape[0], -1)
        X_test_r = X_test.reshape(X_test.shape[0], -1)

    return X_train_r, y_train_enc, X_test_r, y_test_enc, encoder


def calcular_pesos_clase(y_train_enc):
    clases = np.unique(y_train_enc)
    pesos = compute_class_weight(class_weight='balanced', classes=clases, y=y_train_enc)
    return dict(zip(clases, pesos))


def entrenar_modelo(model, X_train, y_train, class_weights, epochs=30, batch_size=128,
                     nombre_modelo='modelo'):
    # Split aleatorio Y ESTRATIFICADO (ver justificación en cnn_modelo.py)
    X_tr, X_val, y_tr, y_val = train_test_split(
        X_train, y_train, test_size=0.15, stratify=y_train, random_state=42
    )

    early_stop = callbacks.EarlyStopping(
        monitor='val_loss', patience=5, restore_best_weights=True
    )

    history = model.fit(
        X_tr, y_tr,
        validation_data=(X_val, y_val),
        epochs=epochs,
        batch_size=batch_size,
        class_weight=class_weights,
        callbacks=[early_stop],
        verbose=2
    )
    return history


def evaluar_modelo(model, X_test, y_test, encoder, nombre_modelo='modelo'):
    y_pred_probs = model.predict(X_test)
    y_pred = np.argmax(y_pred_probs, axis=1)

    acc = accuracy_score(y_test, y_pred)
    precision, recall, f1, _ = precision_recall_fscore_support(
        y_test, y_pred, average='macro', zero_division=0
    )

    print(f"\n{'='*50}")
    print(f"RESULTADOS EN CONJUNTO DE PRUEBA (DS2) - {nombre_modelo}")
    print(f"{'='*50}")
    print(f"Accuracy:           {acc:.4f}")
    print(f"Precision (macro):  {precision:.4f}")
    print(f"Recall (macro):     {recall:.4f}")
    print(f"F1-score (macro):   {f1:.4f}")

    print("\nReporte por clase:")
    print(classification_report(y_test, y_pred, target_names=encoder.classes_,
                                   zero_division=0))

    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=encoder.classes_)
    fig, ax = plt.subplots(figsize=(6, 6))
    disp.plot(ax=ax, cmap='Blues', values_format='d')
    plt.title(f'Matriz de Confusión - {nombre_modelo}')
    plt.tight_layout()
    plt.savefig(f'confusion_matrix_{nombre_modelo}.png', dpi=100)
    print(f"\nMatriz de confusión guardada en confusion_matrix_{nombre_modelo}.png")

    return {'modelo': nombre_modelo, 'accuracy': acc, 'precision': precision,
            'recall': recall, 'f1': f1}


def graficar_historial(history, nombre_modelo='modelo'):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(history.history['loss'], label='train')
    axes[0].plot(history.history['val_loss'], label='val')
    axes[0].set_title(f'Pérdida (Loss) - {nombre_modelo}')
    axes[0].legend()

    axes[1].plot(history.history['accuracy'], label='train')
    axes[1].plot(history.history['val_accuracy'], label='val')
    axes[1].set_title(f'Accuracy - {nombre_modelo}')
    axes[1].legend()

    plt.tight_layout()
    plt.savefig(f'training_history_{nombre_modelo}.png', dpi=100)
    print(f"Historial de entrenamiento guardado en training_history_{nombre_modelo}.png")
```

```python
"""
Modelo MLP (Multi-Layer Perceptron) - Red neuronal básica
============================================================
Días 7-8 del proyecto: Comparación de arquitecturas

Modelo de referencia SIN estructura convolucional ni temporal:
solo capas densas sobre el latido "aplanado". Sirve como piso
comparativo para medir cuánto aportan CNN y LSTM.

Requisitos: pip install tensorflow scikit-learn numpy matplotlib
"""

from tensorflow.keras import layers, models
from ecg_utils import (cargar_datos, preparar_datos, calcular_pesos_clase,
                          entrenar_modelo, evaluar_modelo, graficar_historial)

NOMBRE_MODELO = 'MLP'


def construir_mlp(input_dim, n_clases):
    model = models.Sequential([
        layers.Input(shape=(input_dim,)),
        layers.Dense(256, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.4),

        layers.Dense(128, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.3),

        layers.Dense(64, activation='relu'),
        layers.Dropout(0.2),

        layers.Dense(n_clases, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy',
                    metrics=['accuracy'])
    return model


if __name__ == '__main__':
    print("Cargando datos...")
    X_train, y_train, X_test, y_test = cargar_datos()
    print(f"Train: {X_train.shape[0]} latidos | Test: {X_test.shape[0]} latidos")

    # forma='plana' -> (n, 200), sin dimensión de canal, porque el MLP
    # no distingue posición temporal: cada muestra es solo un vector de features
    X_train, y_train_enc, X_test, y_test_enc, encoder = preparar_datos(
        X_train, y_train, X_test, y_test, forma='plana'
    )
    print(f"Clases (orden): {list(encoder.classes_)}")

    class_weights = calcular_pesos_clase(y_train_enc)

    model = construir_mlp(input_dim=X_train.shape[1], n_clases=len(encoder.classes_))
    model.summary()

    history = entrenar_modelo(model, X_train, y_train_enc, class_weights,
                                 nombre_modelo=NOMBRE_MODELO)
    graficar_historial(history, nombre_modelo=NOMBRE_MODELO)

    resultados = evaluar_modelo(model, X_test, y_test_enc, encoder,
                                   nombre_modelo=NOMBRE_MODELO)

    model.save(f'modelo_{NOMBRE_MODELO.lower()}_arritmias.keras')
    print(f"\nModelo guardado en modelo_{NOMBRE_MODELO.lower()}_arritmias.keras")
```

```python
"""
Modelo LSTM - Modela dependencias temporales
================================================
Días 7-8 del proyecto: Comparación de arquitecturas

A diferencia de la CNN (patrones morfológicos locales), la LSTM
procesa el latido como una secuencia, capturando cómo evoluciona
la señal a lo largo del tiempo dentro de la ventana.

Requisitos: pip install tensorflow scikit-learn numpy matplotlib
"""

from tensorflow.keras import layers, models
from ecg_utils import (cargar_datos, preparar_datos, calcular_pesos_clase,
                          entrenar_modelo, evaluar_modelo, graficar_historial)

NOMBRE_MODELO = 'LSTM'


def construir_lstm(input_shape, n_clases):
    model = models.Sequential([
        layers.Input(shape=input_shape),

        layers.LSTM(64, return_sequences=True),
        layers.Dropout(0.3),

        layers.LSTM(32, return_sequences=False),
        layers.Dropout(0.3),

        layers.Dense(32, activation='relu'),
        layers.Dense(n_clases, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy',
                    metrics=['accuracy'])
    return model


if __name__ == '__main__':
    print("Cargando datos...")
    X_train, y_train, X_test, y_test = cargar_datos()
    print(f"Train: {X_train.shape[0]} latidos | Test: {X_test.shape[0]} latidos")

    # forma='secuencia' -> (n, 200, 1): la LSTM necesita la dimensión temporal
    X_train, y_train_enc, X_test, y_test_enc, encoder = preparar_datos(
        X_train, y_train, X_test, y_test, forma='secuencia'
    )
    print(f"Clases (orden): {list(encoder.classes_)}")

    class_weights = calcular_pesos_clase(y_train_enc)

    model = construir_lstm(input_shape=(X_train.shape[1], 1), n_clases=len(encoder.classes_))
    model.summary()

    # Nota: LSTM es más lenta de entrenar que CNN con la misma cantidad de datos.
    # Si tarda demasiado en tu máquina, reduce epochs o batch_size más pequeño
    # no ayuda aquí -- prueba subir batch_size a 256 para acelerar.
    history = entrenar_modelo(model, X_train, y_train_enc, class_weights,
                                 nombre_modelo=NOMBRE_MODELO)
    graficar_historial(history, nombre_modelo=NOMBRE_MODELO)

    resultados = evaluar_modelo(model, X_test, y_test_enc, encoder,
                                   nombre_modelo=NOMBRE_MODELO)

    model.save(f'modelo_{NOMBRE_MODELO.lower()}_arritmias.keras')
    print(f"\nModelo guardado en modelo_{NOMBRE_MODELO.lower()}_arritmias.keras")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import load_model
from ecg_utils import cargar_datos, preparar_datos, evaluar_modelo

# 1. Cargar datos de prueba una vez
X_train_raw, y_train_raw, X_test_raw, y_test_raw = cargar_datos()

# 2. Definir los modelos a comparar y sus rutas de archivo
modelos_info = [
    {'nombre': 'CNN', 'path': 'modelo_cnn_arritmias.keras', 'forma': 'secuencia'},
    {'nombre': 'MLP', 'path': 'modelo_mlp_arritmias.keras', 'forma': 'plana'},
    {'nombre': 'LSTM', 'path': 'modelo_lstm_arritmias.keras', 'forma': 'secuencia'}
]

todos_resultados = []

print("Comparando modelos en el conjunto de prueba DS2...")

for info in modelos_info:
    if os.path.exists(info['path']):
        # Cargar modelo
        model = load_model(info['path'])

        # Preparar datos en el formato que requiere este modelo específico
        _, _, X_test, y_test, encoder = preparar_datos(
            X_train_raw, y_train_raw, X_test_raw, y_test_raw, forma=info['forma']
        )

        # Evaluar
        res = evaluar_modelo(model, X_test, y_test, encoder, nombre_modelo=info['nombre'])
        todos_resultados.append(res)
    else:
        print(f"Archivo {info['path']} no encontrado. Saltando {info['nombre']}.")

# 3. Crear DataFrame comparativo
df_comp = pd.DataFrame(todos_resultados)
display(df_comp)

# 4. Visualización comparativa
if not df_comp.empty:
    df_melted = df_comp.melt(id_vars='modelo', var_name='Metrica', value_name='Valor')

    plt.figure(figsize=(10, 6))
    sns.barplot(data=df_melted, x='Metrica', y='Valor', hue='modelo')
    plt.title('Comparativa de Arquitecturas - Clasificación de Arritmias')
    plt.ylim(0, 1.05)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
else:
    print("No se encontraron resultados para graficar.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from tensorflow.keras.models import load_model
from ecg_utils import cargar_datos, preparar_datos, evaluar_modelo

# 1. Cargar datos base
X_tr_raw, y_tr_raw, X_ts_raw, y_ts_raw = cargar_datos()

modelos_info = [
    {'nombre': 'CNN', 'path': 'modelo_cnn_arritmias.keras', 'forma': 'secuencia'},
    {'nombre': 'MLP', 'path': 'modelo_mlp_arritmias.keras', 'forma': 'plana'},
    {'nombre': 'LSTM', 'path': 'modelo_lstm_arritmias.keras', 'forma': 'secuencia'}
]

resultados_lista = []

print("Iniciando comparativa de modelos...")

for info in modelos_info:
    if os.path.exists(info['path']):
        print(f"\nEvaluando {info['nombre']}...")
        model = load_model(info['path'])

        # Formatear datos según la arquitectura
        _, _, X_test, y_test, encoder = preparar_datos(
            X_tr_raw, y_tr_raw, X_ts_raw, y_ts_raw, forma=info['forma']
        )

        # Obtener métricas
        m = evaluar_modelo(model, X_test, y_test, encoder, nombre_modelo=info['nombre'])
        resultados_lista.append(m)
    else:
        print(f"[!] No se encontró el archivo para {info['nombre']}")

# 2. Resumen en tabla
df_final = pd.DataFrame(resultados_lista)
display(df_final)

# 3. Gráfico comparativo
if not df_final.empty:
    df_plot = df_final.melt(id_vars='modelo', var_name='Métrica', value_name='Valor')
    plt.figure(figsize=(12, 6))
    sns.barplot(data=df_plot, x='Métrica', y='Valor', hue='modelo')
    plt.title('Comparativa Final: CNN vs MLP vs LSTM')
    plt.ylim(0, 1.0)
    plt.legend(loc='lower right')
    plt.grid(axis='y', alpha=0.3)
    plt.show()
```
